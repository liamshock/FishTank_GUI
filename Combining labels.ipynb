# Convert Labels

If you have several people labelling the same experiment, and they each label a different section of frames,
you can combine all the results easily if you know the frame ranges each person annotated

```python
import numpy as np
import h5py 
import os
```

```python
# main folder path containin the data
folderPath = '/home/liam/Data/returned_labels/'

# the paths for annotators to their labels
liamPath = os.path.join(folderPath, 'liam/liam_100000_200000.h5')
manishaPath = os.path.join(folderPath, 'manisha/Manisha_400000to500000.h5')
thomasPath = os.path.join(folderPath, 'thomas/Thomas_0to100000.h5')
rafaelPath = os.path.join(folderPath, 'rafael/labels.h5')
tatsuoPath = os.path.join(folderPath, 'tatsuo/tatsuo_labels_300000-400000.h5')

# set the annotator ranges
liamRange = [100000, 200000]
manishaRange = [400000, 500000]
thomasRange = [0, 100000]
rafaelRange = [200000, 300000]
tatsuoRange = [300000, 400000]
```

```python
# use one person's data to define the standard arrays that will be updated
with h5py.File(liamPath, 'r') as hf:
    annotated_frames = hf['annotated_frames'][:]
    missing_bps = hf['missing_bps'][:]
    tracks_3D = hf['tracks_3D'][:]
    tracks_imCoords = hf['tracks_imCoords'][:]
```

```python
# load each persons data, placing it in the correct location of the main data

paths = [manishaPath, thomasPath, rafaelPath, tatsuoPath]
franges = [manishaRange, thomasRange, rafaelRange, tatsuoRange]

numPeople = 4

for i in range(4):
    path = paths[i]
    frange = franges[i]
    
    with h5py.File(path, 'r') as hf:
        _annotated_frames = hf['annotated_frames'][:]
        _missing_bps = hf['missing_bps'][:]
        _tracks_3D = hf['tracks_3D'][:]
        _tracks_imCoords = hf['tracks_imCoords'][:]
        
    annotated_frames[frange[0]:frange[1]] = _annotated_frames[frange[0]:frange[1]]
    missing_bps[:, frange[0]:frange[1]] = _missing_bps[:, frange[0]:frange[1]]
    tracks_3D[frange[0]:frange[1]] = _tracks_3D[frange[0]:frange[1]]
    tracks_imCoords[:, frange[0]:frange[1]] = _tracks_imCoords[:, frange[0]:frange[1]]
```

```python
# how many annotated frames
annotated_frames[annotated_frames==True].shape
```

```python
# save the final results

combinedPath = os.path.join(folderPath, 'combined_annotations.h5')
with h5py.File(combinedPath, 'w') as hf:
    hf.create_dataset('annotated_frames', data=annotated_frames)
    hf.create_dataset('missing_bps', data=missing_bps)
    hf.create_dataset('tracks_3D', data=tracks_3D)
    hf.create_dataset('tracks_imCoords', data=tracks_imCoords)
```
